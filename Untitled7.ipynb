```python
from re import sub
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
dataset = pd.read_csv('food2.csv')
x = dataset.iloc[:, 2:8]
y = dataset.iloc[:, -1].map({'Healthy': 1, 'Unhealthy': 0})
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X = sc.fit_transform(x)
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 21)
from xgboost import XGBClassifier
classifier = XGBClassifier(n_estimators = 5, eval_metric = 'logloss', random_state = 21)
classifier.fit(X_train, y_train)
y_pred = classifier.predict(X_test)
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay, classification_report
acc = accuracy_score(y_test, y_pred)
print(f"Accuracy is {round(acc*100, 2)}")
print(f"Misclassification is {round((1-acc)*100, 2)}")
cm = confusion_matrix(y_test, y_pred)
import matplotlib.pyplot as plt
disp = ConfusionMatrixDisplay(confusion_matrix = cm, display_labels = y_test.unique())
disp.plot(cmap = plt.cm.Blues)
plt.title("Food Health Index")
plt.show()
def predict_food_health(sample_features):

  sample_df = pd.DataFrame([sample_features], columns=x.columns)
  scaled_sample = sc.transform(sample_df)

  prediction = classifier.predict(scaled_sample)

  result_label = 'Healthy' if prediction[0] == 1 else 'Unhealthy'
  return result_label


print("\n\n\n\n")
from IPython.display import display
new_data = pd.read_csv('food2_updated.csv')
new_x = new_data.iloc[:, 2:8]
new_x = sc.transform(new_x)
new_data['Predicted_Health'] = new_data.apply(
    lambda row: predict_food_health(row.iloc[2:8].values), axis=1)
display(new_data[['Combo ID', 'Combo Composition', 'Predicted_Health']])
if 'Health_Index' in new_data.columns:
    accuracy = (new_data['Predicted_Health'] == new_data['Health_Index']).mean()
print(f'Prediction Accuracy: {accuracy * 100:.2f}%')
```
